```python
#### extract data from manual annotation script

## import relevant packages
import numpy as np
import os
import matplotlib.pyplot as plt
import glob
import yaml
from tkinter import filedialog as fd
from tkinter import *


def openfiles():
    """
    Script for choosing the manual annotation file to extract data
    eventually will modify to include a choose folder option
    """
    
    print('choose file to extract')
    root = Tk()
    path = fd.askopenfilename()
    root.withdraw()
    filename = os.path.split(path)[1][:-20]
    rootpath = os.path.split(path)[0]
    annot = np.load(path)
    ## returns extracted annotation info, filename (for saving) and directory (for saving)
    return(annot, filename, rootpath)


def processbouts(annot, framerate = 30, minimumboutduration = 0.25, minimumboutgap = 0.25):
    """
    Identifies all the labels included in the dataset and extracts bouts 
    with a minimum gap between bouts and a minimum duration of the bouts
    """
    ### creates a syncfile based on instructed framerate
    videolength = len(annot)*(1/framerate)
    sync = np.arange(0, videolength, 1/framerate)
    
    ## identifies unique annotation labels in data
    labels = np.unique(annot)
    labels = labels[labels != 'nan']
    
    ## extracts the times at which those labels occur and converts in continuous bouts using boutanalysis function
    for x, xx in enumerate(labels):
        behaviortimes = sync[np.where((annot == xx))]
        bouts = boutanalysis(behaviortimes, 0, sync[-1], minimumboutgap, minimumboutduration)
        ## format and stack the data
        labelledbouts = np.hstack((np.full(bouts.shape[0], xx).reshape(-1,1), bouts))
        if x == 0:
            allbouts = labelledbouts
        else:
            allbouts = np.vstack((allbouts, labelledbouts))    
    return allbouts


def boutanalysis(annotbehavior, starttime, endtime, minimumboutgap, minimumboutlength):
    ## extract information that occurs within time window
    behavior = annotbehavior[(annotbehavior > starttime) & (annotbehavior < endtime)]
    
    ## look for start and stop of each bout defined as > 0.05s apart ~2-3 frames
    startindex = np.hstack((0, np.where(np.diff(behavior) > minimumboutgap)[0] + 1))
    endindex = np.hstack((np.where(np.diff(behavior) > minimumboutgap)[0], len(behavior) -1))
    bouts = behavior[np.vstack((startindex, endindex)).T]
    
    ### remove bouts with a duration less than minimumboutlength
    shortidx = np.where(bouts[:,1] - bouts[:,0] < minimumboutlength)
    bouts = np.delete(bouts, shortidx, axis = 0)
    bouts = np.array(bouts, np.float32)
    return bouts


def run_extractbouts(framerate=30, minimumboutduration = 0.1, minimumboutgap = 0.1):
    """ 
    Runs all the scripts together and provides an avenue for modifying fr, boutduration and boutgap
    """
    [annot, filename, rootpath] = openfiles()
    allbouts = processbouts(annot, framerate=framerate, minimumboutduration = minimumboutduration, minimumboutgap = minimumboutgap)
    np.savetxt(rootpath+'/'+filename+'_bouts.csv', allbouts, delimiter = ',', fmt = '%s')
```

```python
run_extractbouts(framerate=30, minimumboutduration=0.1, minimumboutgap = 0.1)
```
