```python
import os

folder_path = r'F:\Loukia-Ruth_CutData\M1116808'
prefix = folder_path[-7:]+'_'
print(prefix)


for filename in os.listdir(folder_path):
    # Remove all repeated prefixes
    new_filename = filename
    while new_filename.startswith(prefix):
        new_filename = new_filename[len(prefix):]
    # Add prefix once
    new_filename = prefix + new_filename
    old_path = os.path.join(folder_path, filename)
    new_path = os.path.join(folder_path, new_filename)
    if os.path.isfile(old_path) and old_path != new_path:
        os.rename(old_path, new_path)

print("Prefix added only once to all files.")
```

```python
import os
import glob

base_path = '/Volumes/Loukia/260306_AxonaThioS_Tiffs'

for filepath in glob.glob(os.path.join(base_path, '**', '*c2.tif'), recursive=True):
    folder = os.path.dirname(filepath)
    filename = os.path.basename(filepath)
    new_filename = filename[:-6] + '_AF488.tif'
    new_path = os.path.join(folder, new_filename)
    os.rename(filepath, new_path)
    print(f"Renamed: {filename}  ->  {new_filename}")

print("Done.")
```

```python
import os
import glob

base_path = '/Volumes/Loukia/260306_AxonaThioS_Tiffs'

for filepath in glob.glob(os.path.join(base_path, '**', '*.tif'), recursive=True):
    folder = os.path.dirname(filepath)
    filename = os.path.basename(filepath)
    if 'ThioS_Axona__' in filename:
        new_filename = filename.replace('ThioS_Axona__', 'ThioS_NLGF_F_')
        new_path = os.path.join(folder, new_filename)
        os.rename(filepath, new_path)
        print(f"Renamed: {filename}  ->  {new_filename}")

print("Done.")
```
