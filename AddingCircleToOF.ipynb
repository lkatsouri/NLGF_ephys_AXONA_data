```python
def process_video(input_video_path, output_video_path, circle_center, circle_diameter):
    # Open the video file
    cap = cv2.VideoCapture(input_video_path)

    # Get the video properties
    frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps = int(cap.get(cv2.CAP_PROP_FPS))

    # Define the codec and create VideoWriter object
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    out = cv2.VideoWriter(output_video_path, fourcc, fps, (frame_width, frame_height))

    # Define circle parameters
    circle_color = (0, 255, 0)  # Green color in BGR
    circle_radius = int(circle_diameter / 2)
    circle_thickness = 2

    # Process each frame
    while True:
        ret, frame = cap.read()
        if not ret:
            break

        # Draw a circle on the frame
        cv2.circle(frame, circle_center, circle_radius, circle_color, circle_thickness)

        # Write the modified frame to the output video
        out.write(frame)

        # Display the frame
        cv2.imshow('Frame', frame)
        if cv2.waitKey(25) & 0xFF == ord('q'):
            break

    # Release video capture and writer objects
    cap.release()
    out.release()

    # Close all OpenCV windows
    cv2.destroyAllWindows()

# Define input video paths
input_video_paths = ['G:/Angela/OF/videos/1118796_20230904_172201317downsampled.mp4',
                     'G:/Angela/OF/videos/1119464_OF_20231130_151251062downsampled.mp4',
                     'G:/Angela/OF/videos/1120158_OF_20240311_123223376downsampled.mp4',
                     'G:/Angela/OF/videos/1119663_OF_20240103_124703275downsampled.mp4']

# Define output video paths
output_video_paths = ['1118796_20230904_172201317downsampled_circle.mp4',
                      '1119464_OF_20231130_151251062downsampled_circle.mp4',
                      '1120158_OF_20240311_123223376downsampled_circle.mp4',
                      '1119663_OF_20240103_124703275downsampled_circle.mp4']

# Define circle center and diameter for each video
circle_centers = [(613, 490),  # Center for the first video
                  (625, 492),     # Center for the second video
                  (633, 492),     # Center for the third video
                  (633, 496)]     # Center for the fourth video

circle_diameters = [220.41,  # Diameter for the first video
                    220.41,   # Diameter for the second video
                    220.41,   # Diameter for the third video
                    220.41]   # Diameter for the fourth video

# Process each video
for i in range(len(input_video_paths)):
    process_video(input_video_paths[i], output_video_paths[i], circle_centers[i], circle_diameters[i])
```
