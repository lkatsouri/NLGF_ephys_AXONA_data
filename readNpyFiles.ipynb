```python
"""To find the first insance of STart analysis and check if there is one"""
basepath = 'G:/Angela/Light-dark_box/downsampled_videos/'

import os
os.listdir(basepath)
directory = os.listdir(basepath)

annotpath = []
for i, ii in enumerate(directory):
    if ii[-3:] == 'npy':
        annotpath = np.append(annotpath, os.path.join(basepath, ii))  # this is to append the list with all the videos
        annotpath_list = annotpath.tolist()

for f, ff in enumerate(annotpath_list):
    annot = np.load(annotpath_list[f])
    startidx = np.where(annot == 'Start Analysis')[0][0]
    print(ff, startidx)
```
