```python
import pandas as pd
import os

def analyze_y_maze(file_path, save_path=None, sheet_name=0):
    # Load Excel file
    df = pd.read_excel(file_path, sheet_name=sheet_name, header=0)

    # Extract experiment IDs and arm entry data
    experiment_ids = df.iloc[:, 0]  # First column as experiment ID
    arm_data = df.iloc[:, 1:]  # Ignore first column for calculations

    # Initialize result storage
    results = []

    # Iterate over each row (each experiment)
    for index, row in arm_data.iterrows():
        experiment_id = experiment_ids.iloc[index]
        sequence = row.dropna().astype(str).tolist()  # Get arm entries

        # Calculate possible triplets
        possible_triplets = max(len(sequence) - 2, 0)

        # Initialize counters
        sap_count = 0  # Spontaneous Alternation Performance
        sar_count = 0  # Same Arm Returns
        aar_count = 0  # Alternate Arm Returns

        # Iterate over overlapping triplets
        for i in range(len(sequence) - 2):
            triplet = sequence[i:i+3]

            if len(set(triplet)) == 3:  
                sap_count += 1  # All different
            elif triplet[0] == triplet[1] or triplet[1] == triplet[2]:  
                sar_count += 1  # First & second OR second & third are the same
            elif triplet[0] == triplet[2] and triplet[0] != triplet[1]:  
                aar_count += 1  # First & third are the same, second is different

        # Calculate percentages (handle division by zero)
        sap_percentage = (sap_count / possible_triplets) * 100 if possible_triplets > 0 else 0
        sar_percentage = (sar_count / possible_triplets) * 100 if possible_triplets > 0 else 0
        aar_percentage = (aar_count / possible_triplets) * 100 if possible_triplets > 0 else 0

        # Store results
        results.append({
            "Experiment ID": experiment_id,
            "Total Arm Entries": len(sequence),
            "Possible Triplets": possible_triplets,
            "SAP": sap_count,
            "SAR": sar_count,
            "AAR": aar_count,
            "% SAP": round(sap_percentage, 2),
            "% SAR": round(sar_percentage, 2),
            "% AAR": round(aar_percentage, 2)
        })

    # Convert results to DataFrame
    result_df = pd.DataFrame(results)

    # Define save path
    if save_path:
        os.makedirs(save_path, exist_ok=True)  # Ensure directory exists
        result_file = os.path.join(save_path, "y_maze_Manual_results_12moNLGF.xlsx")
    else:
        result_file = "y_maze_Manual_results_12moNLGF.xlsx"  # Default save in current directory

    # Save results to the specified path
    result_df.to_excel(result_file, index=False)

    print(f"Results saved at: {result_file}")

    return result_df
```
